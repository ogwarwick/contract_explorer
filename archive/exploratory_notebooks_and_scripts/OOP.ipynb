```python
class Fruit:
    def __init__(self,name,price,quantity):
        self.name = name
        self.price = price
        self.quantity = quantity
    def expected_revenue(self):
        return (self.price * self.quantity)

f1 = Fruit("Banana",0.25,1000)
revenue = f1.expected_revenue()
print(revenue)
```

```python
class Stock:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def quantity(self):
        return self._quantity

    def quantity(self, new_quantity):
        if new_quantity < 0:
            raise ValueError("Quantity is negative")
        self._quantity = new_quantity

    def expected_revenue(self):
        return (self.price * self.quantity)

class Fruit(Stock):
    def __init__(self,name,price,quantity,origin):
        super().__init__(name,price,quantity)
        self.origin = origin
    def get_tag(self):
        return f"{self.name} - Grown in {self.origin}"

class Dairy(Stock):
    def __init__(self,name,price,quantity,lid_top):
        super().__init__(name,price,quantity)
        self.lid_top = lid_top
    def get_tag(self):
        return f"{self.name} - Lid type {self.lid_top}"

orange = Fruit("Orange",0.75,100, origin="Spain")
cravendale_green = Dairy("Cravendale",3,100,lid_top = "Green")
banana = Fruit("Banana",0.25,100, origin="Colombia")

print(orange.expected_revenue()+cravendale_green.expected_revenue())
```
